**CO4: Group Operations and Time Series Analysis**

Pandas provides powerful tools for grouping, aggregating, and analyzing time-based data. In this section, Pandas is used to summarize and analyze the web traffic dataset using group operations, convert dates to datetime format, perform time-based selection, and apply resampling and frequency changes to analyze web traffic trends.

```python
# Group by Traffic Source and Engagement Category
g1 = df.groupby(
    ["Traffic Source", "Engagement Category"]
)["Page Views"].agg(
    ["count", "mean"]
).round(2)

print(g1)
```

```python
# Average conversion rate by traffic source
conversion = df.groupby(
    "Traffic Source"
)["Conversion Rate"].agg(
    ["count", "mean"]
).rename(
    columns={"mean": "average_conversion_rate"}
)

print(conversion.round(3))
```

```python
# Rank traffic sources by average conversion rate
conversion_rank = conversion.sort_values(
    "average_conversion_rate",
    ascending=False
)

print("Traffic Sources Ranked by Conversion Rate:")
print(conversion_rank.round(3))
```

```python
# Descriptive statistics by traffic source
source_stats = df.groupby("Traffic Source").agg({
    "Page Views": ["mean", "median", "std"],
    "Session Duration": ["mean", "median", "std"],
    "Bounce Rate": ["mean", "median"],
    "Time on Page": ["mean", "median"],
    "Conversion Rate": ["mean", "median"]
}).round(3)

source_stats
```

```python
# Create a copy for time-series analysis
time_df = df.copy()

# Create a synthetic hourly date-time column for demonstration
time_df["Date"] = pd.date_range(
    start="2026-01-01",
    periods=len(time_df),
    freq="h"
)

# Convert Date column to datetime format
time_df["Date"] = pd.to_datetime(time_df["Date"])

time_df[["Date", "Traffic Source", "Page Views"]].head()
```

```python
# Set 'Date' as the index to enable datetime slicing and resampling
time_df = time_df.set_index("Date")

# Select data from January 1 to January 2
jan_data = time_df.loc["2026-01-01":"2026-01-02"]

# Resample hourly data into daily data
daily_traffic = time_df.resample("D").agg({
    "Page Views": "mean",
    "Session Duration": "mean",
    "Bounce Rate": "mean",
    "Time on Page": "mean",
    "Conversion Rate": "mean"
}).round(3)

daily_traffic.head()
```

```python
# Compare different frequencies

hourly_views = time_df["Page Views"].resample("h").mean()
daily_views = time_df["Page Views"].resample("D").mean()
weekly_views = time_df["Page Views"].resample("W").mean()

print("Hourly data:")
print(hourly_views.head())

print("\nDaily data:")
print(daily_views.head())

print("\nWeekly data:")
print(weekly_views.head())
```

**CO5: Data Visualization**

Matplotlib and Seaborn provide powerful tools for creating informative visualizations from structured data. In this section, these libraries are used to visually analyze the web traffic dataset through bar plots, distribution plots, and heatmaps, highlighting patterns and relationships among traffic sources and website performance metrics.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Traffic Source",
    y="Page Views",
    errorbar=None
)

plt.title("Average Page Views by Traffic Source")
plt.xlabel("Traffic Source")
plt.ylabel("Average Page Views")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Traffic Source",
    y="Session Duration"
)

plt.title("Session Duration by Traffic Source")
plt.xlabel("Traffic Source")
plt.ylabel("Session Duration")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))

conversion_rate = df.groupby("Traffic Source")["Conversion Rate"].mean()

plt.pie(
    conversion_rate,
    labels=conversion_rate.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Average Conversion Rate by Traffic Source")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="Traffic Source",
    y="Page Views",
    hue="Engagement Category",
    errorbar=None
)

plt.title("Average Page Views by Traffic Source and Engagement Category")
plt.xlabel("Traffic Source")
plt.ylabel("Average Page Views")
plt.xticks(rotation=45)
plt.legend(title="Engagement Category")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))

correlation = df[
    [
        "Page Views",
        "Session Duration",
        "Bounce Rate",
        "Time on Page",
        "Previous Visits",
        "Conversion Rate"
    ]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Web Traffic Metrics")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="Traffic Source",
    y="Bounce Rate"
)

plt.title("Bounce Rate Distribution by Traffic Source")
plt.xlabel("Traffic Source")
plt.ylabel("Bounce Rate")

plt.tight_layout()
plt.show()
```

## Conclusion

The Web Traffic Source Analysis project was carried out to understand website traffic patterns, user engagement, and conversion performance across different traffic sources.

### Key Findings

- **Organic traffic** was the most common traffic source and generated the highest average number of page views.
- **Referral traffic** recorded the longest average session duration, indicating higher user engagement.
- **Referral traffic** also had the lowest average bounce rate, suggesting that visitors from this source were more likely to explore the website.
- **Social traffic** recorded the highest average time on page, indicating greater content engagement.
- **Referral traffic** achieved the highest average conversion rate among the different traffic sources.

### Overall Insight

The analysis shows that different traffic sources contribute differently to website performance. **Organic traffic** is particularly effective in generating traffic volume and page views, while **referral traffic** performs strongly in terms of user engagement and conversions. **Social traffic** also demonstrates strong engagement through higher time spent on the website. These findings can help identify the most effective traffic channels and support data-driven decisions for improving website performance and marketing strategies.
